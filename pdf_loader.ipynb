```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("2017-attention-is-all-you-need.pdf")
file_path.exists()
```

```python
# To convert file path in string
file_path.as_posix()
```

```python
pypdf_loader = PyPDFLoader(file_path=file_path.as_posix(),
                           mode = 'page')
```

```python
# load the documents
documents = pypdf_loader.load()
```

```python
len(documents)
```

```python
# first page txt context
print(documents[0].page_content)
```

```python
pp(documents[1].metadata)
```

```python
from langchain_community.document_loaders.parsers import TesseractBlobParser,RapidOCRBlobParser
```

```python
# For extracting images
pypdf_image_loader = PyPDFLoader(file_path=file_path.as_posix(),
                                 mode='page',
                                 extract_images=True,
                                 images_parser=RapidOCRBlobParser(),
                                 images_inner_format='html-img')
```

```python
# load documents with images
documents_with_images = pypdf_image_loader.load()
```

```python
len(documents_with_images)
```

```python
print(documents_with_images[2].page_content)
```

```python
# We have to use pdfminer to extract images
from langchain_community.document_loaders import PDFMinerLoader
```

```python
pdf_miner = PDFMinerLoader(file_path=file_path.as_posix(),
                           mode='page',
                           extract_images=True,
                           images_parser=RapidOCRBlobParser(),
                           images_inner_format='html-img')
```

```python
documents_with_images_miner = pdf_miner.load()
```

```python
print(documents_with_images[2].page_content)
```
